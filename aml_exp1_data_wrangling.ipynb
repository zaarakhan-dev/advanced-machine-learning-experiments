## 1. Importing Libraries
Import essential libraries for data manipulation and preprocessing:
* **`pandas`**: Used for creating, structuring, and transforming DataFrames.
* **`MinMaxScaler` (from `sklearn.preprocessing`)**: Used to scale numerical features into a standardized range $[0, 1]$.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

## 2. Dataset Creation
Generate a synthetic student dataset (`df1`) with 20 records containing:
* **`ID`** and **`Name`**: Unique identifiers and student names.
* **`Age`** and **`Marks`**: Continuous numerical features intentionally populated with missing values (`None`/`NaN`) to simulate real-world data quality issues.

```python
data1 = {
    'ID': range(1,21),
    'Name': [
        'Aman', 'Riya', 'Rahul', 'Priya', 'Karan',
        'Neha', 'Arjun', 'Simran', 'Rohit', 'Pooja',
        'Vikas', 'Anjali', 'Nikhil', 'Sneha', 'Varun',
        'Kavita', 'Mohit', 'Isha', 'Aditya', 'Sakshi'
    ],

    'Age':[
        21,20,19,18,21,
        22,24,None,21,None,
        19,20,21,22,20,
        21,20,None,18,20
    ],

    'Marks':[
         85, 90, 78, None, 88,
        76, 95, 82, None, 91,
        69, 87,None, 96, 80,
        None, 89, 75, 92, 84
    ]
}

df1=  pd.DataFrame(data1)
print(df1)
```

## 3. Data Reshaping (`pd.melt`)
Unpivot the DataFrame from **wide format** to **long format** using `pd.melt()`:
* **Identifier variables (`id_vars`)**: Kept fixed (`ID`, `Name`).
* **Value variables (`value_vars`)**: Stuffed into key-value pairs (`Features` and `Value`).

> **Why reshape?** Long formats are commonly required for certain statistical models, Seaborn plotting, and standardized pipeline inputs.

```python
reshaped_data = pd.melt(
    df1,
    id_vars=['ID','Name'],
    value_vars=['Age','Marks'],
    var_name='Features',
    value_name='Value'
)
print(reshaped_data)
```

## 4. Data Filtering (Conditional Slicing)
Filter the dataset using boolean indexing to retain only adult students:
$$\text{Condition: } \text{Age} > 18$$

* Excludes students aged 18 or younger.
* Automatically excludes records where `Age` is missing (`NaN`), as boolean comparisons against `NaN` evaluate to `False`.

```python
filtered_data = df1[df1['Age']>18]
print(filtered_data)
```

## 5. Dataset Merging (`pd.merge`)
Create a secondary DataFrame (`df2`) containing academic metadata (`Course`, `City`, `Department`) and combine it with `df1`:
* **Join Key**: Inner join performed on the primary key `ID`.
* **Result**: A consolidated view combining student performance data with institutional details.

```python
data2 ={
    'ID': range(1, 21),

    'Course': [
        'MCA', 'BCA', 'MCA', 'BCA', 'B.Tech',
        'BCA', 'B.Tech', 'BCA', 'MCA', 'BCA',
        'MCA', 'BCA', 'B.Tech', 'BCA', 'MCA',
        'BCA', 'MCA', 'BCA', 'MCA', 'B.Tech'
    ],

    'City': [
        'Delhi', 'Gurgaon', 'Noida', 'Delhi', 'Faridabad',
        'Gurgaon', 'Delhi', 'Noida', 'Delhi', 'Gurgaon',
        'Noida', 'Delhi', 'Faridabad', 'Gurgaon', 'Delhi',
        'Noida', 'Delhi', 'Faridabad', 'Gurgaon', 'Delhi'
    ],

    'Department':[
        'SOET', 'SOET', 'SOET', 'SOET', 'SOET',
         'SOET', 'SOET', 'SOET', 'SOET', 'SOET',
        'SOET', 'SOET', 'SOET', 'SOET', 'SOET',
         'SOET', 'SOET', 'SOET', 'SOET', 'SOET'
    ]
}
df2 = pd.DataFrame(data2)

merged_dataset = pd.merge(df1,df2, on='ID')

print(merged_dataset)
```

## 6. Handling Missing Data (Mean Imputation)
Address missing values (`NaN`) in the numerical columns using **Mean Imputation**:
$$\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i$$

* Replaces missing values in `Age` and `Marks` with their respective column averages.
* Prevents data loss compared to dropping rows (`dropna()`), preserving the total sample size ($N = 20$).

```python
df1['Age'] = df1['Age'].fillna(df1['Age'].mean())
df1['Marks'] = df1['Marks'].fillna(df1['Marks'].mean())

print(df1)
```

## 7. Feature Normalization (`MinMaxScaler`)
Scale the numerical features (`Age` and `Marks`) to a uniform range $[0, 1]$ using Min-Max scaling:
$$X_{\text{scaled}} = \frac{X - X_{\min}}{X_{\max} - X_{\min}}$$

* **Why normalize?** Prevents features with larger scales (such as `Marks`, ranging up to 96) from dominating features with smaller scales (such as `Age`, ranging up to 24) in distance-based machine learning algorithms (e.g., KNN, SVM, K-Means).

```python
scaler = MinMaxScaler()
df1[['Age','Marks']] = scaler.fit_transform(df1[['Age','Marks']])
print(df1)
```
